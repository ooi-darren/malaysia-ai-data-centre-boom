# 01: Why Malaysia, Why Johor

**Business question**: In under a decade, a stretch of former oil-palm land on the Johor-Singapore border has become one of Southeast Asia's fastest-growing data centre markets. Was this a deliberate policy bet, an accident of geography, or both, and what actually changed to make it possible?

## The scale of the shift

Live data centre capacity in Malaysia grew from **55MW in 2019 to 784MW by the third quarter of 2025**, a 55.6% compound annual growth rate. **[PUBLIC, Bank Negara Malaysia Q3 2025 Quarterly Bulletin box article "From Bytes to Bucks: The Economics of Data Centres in Malaysia," Chart 2, sourced to DC Byte and BNM's own estimates]** By end-2026, national live capacity is estimated to reach **2,055MW**, roughly doubling again, with a further 3,500MW in the pipeline beyond that. **[PUBLIC, JLL Malaysia data centre market report]** Johor alone accounts for an estimated 850MW already built, 1,800MW under construction, and 2,700MW in the pipeline, roughly 60% of the projected national total by 2030. **[PUBLIC, JLL Malaysia data centre market report; JLL's terms of use restrict reproduction of its content, so this figure is cited as a specific attributed statistic, not reproduced from the report itself, see the compliance note in this notebook's Sources cell]**

Data centres also went from a minor contributor to Malaysia's foreign direct investment to its single largest one in three years: their share of net FDI inflows rose from **4% in 2022, to 37% in 2023, to 51% in 2024**, overtaking manufacturing. **[PUBLIC, Bank Negara Malaysia Q3 2025 Bulletin, Chart 3, sourced to the Department of Statistics Malaysia and BNM]**

## Why Johor specifically

Bank Negara Malaysia's own analysis names three factors that pulled this investment toward Malaysia, and Johor in particular: strategic location (proximity to undersea communications cables and to Singapore, whose own market is capacity-constrained), resource availability (energy, water, land, and skilled labour, plus faster grid connection timelines than neighbouring markets), and a deliberate incentive policy from MIDA and the Malaysia Digital Economy Corporation (MDEC). **[PUBLIC, Bank Negara Malaysia Q3 2025 Bulletin, Diagram 1]**

The Johor-Singapore Special Economic Zone (JS-SEZ), formalised in 2025, sits on top of a pre-existing structural advantage: industrial land and power in Johor cost a fraction of Singapore's, at a time when Singapore's own data centre market was, by design, growing slowly (see Notebook 11 for the direct comparison). TNB itself cut its typical grid-connection timeline for data centres from 36 months to as little as 12 months under a "Green Lane Pathway" introduced in 2026, a concrete example of policy actively shortening the path from land purchase to live capacity. **[PUBLIC, The Star, 28 May 2026]**

This project's later notebooks complicate a simple "unrestrained gold rush" reading of this growth, however. As Notebook 3 details, Malaysia's government has been quietly filtering this pipeline since around mid-2024, not merely encouraging it.

## Explain It Simply

Big technology companies need physical buildings full of computer servers to run things like AI models, cloud storage, and streaming video: these are data centres, and they need huge amounts of land, electricity, and water to run. Singapore, the obvious place to build them in this region, ran out of easy room for more years ago and has deliberately kept a tight lid on new ones since 2019. Malaysia, and Johor state specifically, sat right next door with cheaper land, more available power, and a government actively trying to attract this investment, so a huge wave of new data centre construction has landed just across the border since around 2019, and especially since 2022. This project asks whether that wave has actually turned into real economic activity yet, or whether it is still mostly a pipeline of announcements and construction sites.

## Glossary

- **Data centre**: a purpose-built facility housing servers and networking equipment, used to run cloud computing, AI, and internet services.
- **Hyperscaler**: a very large cloud provider (Amazon Web Services, Microsoft Azure, Google Cloud, and similar) that builds and operates data centres at massive scale.
- **JS-SEZ (Johor-Singapore Special Economic Zone)**: a 2025 bilateral economic zone agreement between Malaysia and Singapore covering the Johor border region, intended to ease cross-border investment.
- **MIDA / MDEC**: the Malaysian Investment Development Authority and Malaysia Digital Economy Corporation, the two government bodies responsible for approving and promoting this investment.

## Why Is This Happening?

**Why did this investment wave land specifically in Johor rather than spreading evenly across Malaysia, or going to a different country entirely?** Data centres are unusually sensitive to a narrow set of physical constraints: proximity to internet backbone infrastructure, reliable and abundant electricity, water for cooling, and large contiguous parcels of land, all of which need to sit within a short, low-latency distance of the customers the facility will actually serve, in this case, Singapore's dense concentration of regional corporate and financial activity. Johor is the only location that offers cheap land and power within that latency radius of Singapore, which is itself unable to expand quickly under its own tighter environmental rules (Notebook 11). That combination of hard geographic constraint and a policy vacuum next door, filled deliberately by MIDA and MDEC incentives, explains why this specific stretch of land became the region's overflow valve rather than any other part of Malaysia or Southeast Asia.
