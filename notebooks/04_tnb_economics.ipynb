## A near-tripling in data centres' share of national demand

Malaysia's own government planning document (JPPPET No 1/2026) forecasts peak electricity demand rising from **21.3GW (2026) to 33.5GW (2035)**, with data centres' share of total electricity demand rising from **7% to 31%** over the same period, equivalent to 73,274 GWh of annual data centre demand by 2035. **[PUBLIC, written Parliamentary answer from the Ministry of Energy Transition and Water Transformation, dated 6 July 2026, responding to a question from MP Onn Abu Bakar]** This is a government planning record, not a TNB-copyrighted company disclosure, and carries no reproduction restriction.

At the 2026 end of that forecast, 36 data centres were already supplying roughly 4.5GW of demand, with a further 23 projects (3.8GW) under construction, broadly consistent with the utilisation-gap figures in Notebook 3 from a slightly earlier snapshot.

## What it costs TNB to build for this

Under its current Regulatory Period (RP4, January 2025 to December 2027), TNB has an allowed capital expenditure of **RM42.821 billion**, split into RM26.554 billion base capex and RM16.267 billion contingent capex, with the regulatory rate of return held at 7.3%, unchanged from the prior period. **[PUBLIC, SERC/ACCCIM research note, Bernama, and The Edge, all consistently reporting the same Energy Commission-set figure]** This is a **three-year regulatory allocation, not an annual spend**; press coverage that presents "RM43 billion" without that context risks implying a much larger single-year outlay than what was actually approved.

The tariff base was raised 14.2% (from 39.95 to 45.62 sen/kWh) as part of financing this capex programme, later fixed at 45.4 sen/kWh for the remainder of RP4. There is no separate "data centre tariff": large facilities are billed under the existing ultra-high-voltage industrial category, which RP4 restructured from a single flat Maximum Demand charge into separate Capacity and Network Charges. The real cost impact of this restructuring, per one industry consultancy: a 100MW facility now pays an estimated **US$15-20 million more per year**, before fuel surcharges. **[Reuters, 1 July 2025, quoting Sprint DC Consulting]**

Separately, TNB introduced a "Green Lane Pathway" in 2026 that cuts typical grid-connection timelines for data centres from 36 months to as little as 12 months, a genuine operational commitment behind the demand growth shown above, not just a pricing change. **[PUBLIC, The Star, 28 May 2026]**

## Explain It Simply

TNB, Malaysia's electricity company, is planning to spend about RM43 billion over three years upgrading the grid, partly to keep up with data centres, which the government's own planners expect to go from about 7% of all the electricity Malaysia uses today to nearly a third by 2035. There's no special "data centre price" for electricity: big facilities just pay under the same large-industrial-user pricing as any other heavy power user, though a recent pricing restructure means big facilities now pay noticeably more than they used to. TNB has also promised to connect new data centres to the grid faster, cutting a process that used to take three years down to as little as one.

## Glossary

- **Regulatory Period (RP)**: a multi-year cycle (currently RP4, 2025-2027) under which Malaysia's Energy Commission sets TNB's allowed revenue, capital spending, and tariffs.
- **sen/kWh**: Malaysian cents per kilowatt-hour, the standard unit for electricity tariffs.
- **Maximum Demand (MD) charge**: a tariff component based on a customer's peak power draw, historically billed as a single flat charge, now split into separate capacity and network components under RP4.

## Why Is This Happening?

**Why would TNB restructure its tariff specifically around the time data centres arrived at scale?** Large, constant-load industrial customers like data centres draw power very differently from households or typical factories, close to their maximum demand around the clock, rather than in daily peaks and troughs. A flat Maximum Demand charge undercharges this kind of customer relative to the grid capacity it actually reserves, since it doesn't distinguish between a customer who occasionally spikes to their contracted maximum and one who sits there permanently. Splitting the charge into separate capacity and network components lets TNB price the two kinds of customers more accurately, capturing more of the cost data centres genuinely impose on grid planning, which is also the direct funding source for the RM42.8 billion RP4 capex programme referenced above.
