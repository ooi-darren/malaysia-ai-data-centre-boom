## Four real, dated transactions

**EcoWorld sold 49.6 acres at its QUANTUM Edge development in Kulai to KNBDC Malaysia Five Sdn Bhd, a hyperscale data centre specialist, for RM280.8 million** (roughly RM5.66 million per acre), announced around 24 April 2026. EcoWorld's own disclosure states this single sale is expected to lift its **FY2027 profit by RM104-147 million, an 18-26% increase**, a rare case where a listed landowner directly quantifies how much one data centre land sale is worth to its own bottom line. **[PUBLIC, EcoWorld press release, corroborated by The Star, Bernama, and EdgeProp]**

**Princeton Digital Group acquired 31 acres at Sedenak Tech Park from JLand Group**, announced as part of a **RM2 billion total investment (roughly USD450 million) for a planned 150MW campus**, a figure describing total planned investment, not a land price alone, so it is not directly comparable to the two land-only sales above. **[PUBLIC, Princeton Digital Group press release]**

**UEM Sunrise sold two parcels in Iskandar Puteri to an undisclosed "leading global data centre player" for RM144.9 million.** **[PUBLIC, UEM Sunrise announcement]**

**SIPP Power, a YTL Power subsidiary, secured 145 acres at Sedenak Tech Park West, with an option on a further 400 acres**; the price was not disclosed. **[PUBLIC transaction, price undisclosed]** Separately, **Yondr Group secured financing of over USD900 million for a 72.8-acre Sedenak campus**, a financing figure rather than a land price. QUANTUM Edge in Kulai, the site of the EcoWorld sale above, now hosts Microsoft, Princeton Digital Group, and the new KNBDC buyer together, a useful, concrete example of the co-location density this corridor has reached in a short time.

## Explain It Simply

Before a single server gets plugged in, a data centre company first needs to buy land, and the people who owned that land beforehand are cashing in. A palm-oil-adjacent stretch of Johor near the Singapore border has turned into some of the most valuable industrial land in the country almost overnight: one developer, EcoWorld, sold less than 50 acres for RM280.8 million and says that single sale alone will boost its profit by up to a quarter next year. This is one of the clearest, most concrete ways this boom is creating real, already-realised value for somebody, not just future promises, landowners who happened to hold the right parcels are being paid real money today.

## Glossary

- **Land bank**: land a developer holds for future development, whose value can rise sharply if demand for that location increases unexpectedly.
- **Campus** (in a data centre context): a large site designed to host multiple data centre buildings, sometimes built out in phases over several years.

## Why Is This Happening?

**Why would a single land sale move a large listed developer's profit by up to a quarter?** EcoWorld's core business is residential and mixed-use property development, where individual land parcels are typically developed and sold gradually over years at ordinary property-development margins. A hyperscale data centre buyer, by contrast, is not price-sensitive to typical property-development economics in the same way, it needs a specific location (within the JS-SEZ, close to Singapore, with power and connectivity already discussed in Notebook 1) urgently enough that it will pay well above ordinary industrial land rates to secure it. A single opportunistic sale at that premium can be large enough, relative to a developer's normal annual profit, to move headline profit guidance meaningfully, exactly what EcoWorld's own disclosure shows.
