## A genuinely new disclosure, not a multi-year trend yet

Before this year, YTL Power's data centre activity was folded inside a broader "Investment holding activities" segment; the quarterly report for the period ended 31 December 2025 still shows only four reportable segments, with no discrete Data Center line. **Only with the FY2026 annual filing (for the year ended 30 June 2026, filed 20 August 2026) did the company formally break out a fifth segment, "Data Center,"** restating the FY2025 comparative to match. **[PUBLIC, YTL Power International FY2026 Annual Interim Report, Note A8, read directly from the primary filing]**

This means a clean multi-quarter trend the way this portfolio built for other companies (see, for example, project 010's Teleport or ADE notebooks) does not exist yet for YTL Power's data centre business; only a FY2026-vs-FY2025 annual comparison and a Q4-vs-Q4 comparison are available on an audited, comparable basis.

## The real numbers, and the tension sitting right next to them

Data Center segment external revenue rose from **RM124.1 million (FY2025, restated) to RM867.1 million (FY2026)**, with segment profit before tax rising from **RM32.1 million to RM345.5 million**. In the fourth quarter alone, Data Center revenue was **RM459.4 million** against **RM57.6 million** a year earlier. **[PUBLIC, YTL Power International FY2026 Annual Interim Report, Note A8]**

Put in proportion, Data Center is still a small slice of YTL Power's overall business, about **3.9% of FY2026 group revenue**, but an outsized **14.4% of FY2026 group profit before tax**, a small-revenue-base, high-profit-contribution pattern worth stating explicitly rather than quoting the eye-catching growth rate alone.

The genuinely important context this project found: **YTL Power's group-wide profit before tax actually fell 27.6% in FY2026** (RM2,398.1 million versus RM3,310.4 million in FY2025), driven by a 47% collapse in the Power Generation segment's margins. The Data Center segment's roughly tenfold profit growth happened *while the rest of the group's profitability declined sharply*, a real tension worth stating plainly rather than presenting the data centre growth story in isolation. **[PUBLIC, same filing, YoY commentary]**

On capacity, YTL Power states it has **invested RM10.6 billion in its data centre business to date**, working toward a green data centre park capacity of **1,200MW, of which 298MW is already contracted**, co-powered by an adjacent 500MW solar facility (215MWac already built) at Kulai Young Estate. **[PUBLIC, same filing, Note B3 Prospects section, direct quote]**

## Explain It Simply

YTL Power is the one company in this whole project that actually reports its data centre business as its own separate, audited line item, and only started doing so this year. That business grew its profit roughly tenfold in a year, which sounds spectacular, but it's happening off a small starting base: it's still under 4% of YTL Power's total revenue, even though it now contributes a meaningfully larger 14% of total profit. What makes this genuinely interesting rather than simply a good-news story: while data centres were booming for YTL Power, the rest of its business, mainly traditional power generation, had a much worse year, and overall company profit actually fell by more than a quarter. Data centres are a real bright spot inside a company whose other main business is currently struggling, not proof the whole company is thriving.

## Glossary

- **Segment reporting**: the accounting practice of breaking a company's financial results into distinct business lines (segments) rather than reporting only a single consolidated total.
- **PBT (Profit Before Tax)**: profit after operating and financing costs but before tax is deducted, often used to compare segments on a like-for-like basis regardless of differing tax treatments.
- **Restated comparative**: a prior period's figures recalculated to match a new segment structure, so that a year-over-year comparison remains meaningful after a reporting change.

## Why Is This Happening?

**Why would YTL Power only now break out Data Center as its own segment, after presumably building this business for years?** Companies typically create a new reportable segment once a business line crosses a materiality threshold, when its revenue, profit, or strategic importance to the group becomes large enough that lumping it into a catch-all category like "Investment holding activities" would obscure meaningful information from investors. The FY2026 jump to RM867.1 million revenue and RM345.5 million profit before tax, against a Data Center segment that barely registered the year before, is exactly the kind of inflection point that typically triggers this accounting change: this project's own evidence is consistent with Data Center only recently becoming large enough, in the company's own judgement, to warrant separate disclosure.
