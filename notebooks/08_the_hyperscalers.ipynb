## Explain It Simply

Six of the world's biggest tech companies have all announced they're investing billions of dollars in Malaysia, adding up to over $23 billion in pledges. These numbers are real announcements, not made up, but they're a different kind of real from the numbers in earlier notebooks: a Malaysian company like TNB or YTL Power has to file audited reports that can be checked and cross-referenced, but a foreign tech company's investment pledge is just a promise with no equivalent public paper trail in Malaysia to verify against. Some of these pledges stretch out to 2037 or 2038, more than a decade away, and nobody outside these companies currently has a public way to check how much of the promised money has actually shown up yet versus how much is still just a plan.

## Glossary

- **Capex (capital expenditure) pledge**: a company's public commitment to spend a certain amount building physical infrastructure, typically over a multi-year period.
- **Disbursement**: the actual deployment of pledged funds, as distinct from the pledge itself.

## Why Is This Happening?

**Why would a hyperscaler announce a large, multi-year pledge with no obligation to report progress against it?** A public investment pledge functions partly as a genuine capital-allocation decision and partly as a signal, to the host government (supporting continued policy favour and land/incentive access), to competitors (staking a claim to a strategically important market before rivals do), and to the hyperscaler's own investors (demonstrating growth in AI infrastructure capacity). None of those purposes requires the same disclosure discipline as a Malaysia-listed company's regulatory filing, since the pledge is made under the hyperscaler's home-market reporting rules (US securities law, in most of these cases), not Malaysia's. This asymmetry, rigorous local disclosure for Malaysian companies, unaudited pledges for foreign hyperscalers, is a structural feature of how this whole boom gets reported, not a flaw specific to any one company.
