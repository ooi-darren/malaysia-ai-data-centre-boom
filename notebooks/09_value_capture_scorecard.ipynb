## Resolving a comparison this project deliberately does not make

Two large, frequently-quoted figures describe Malaysia's data centre investment: **MIDA's cumulative approved investment, RM385.7 billion (2021 to 1H2026)**, and a much smaller, separately-sourced figure, **Bank Negara Malaysia's own staff estimate of data centres' direct contribution to real GDP, RM14.1 billion in 2025**. These look, at first glance, like a "gap" worth highlighting. They are not the same kind of number: MIDA's figure is a **cumulative, multi-year approval pipeline**, investment applications the government has approved, not money that has necessarily been spent; BNM's figure is a **single year's contribution to GDP**, an entirely different unit of measurement, and explicitly labelled by BNM itself as its staff's own initial estimate, not an audited final statistic. Both figures come from the same Bank Negara Malaysia bulletin, which presents them separately and does not compare them directly. **[PUBLIC, Bank Negara Malaysia Q3 2025 Quarterly Bulletin box article]** This notebook follows BNM's own lead and reports both figures on their own terms, rather than manufacturing a "gap" narrative between two numbers that were never meant to be compared.

## The value-capture picture, layer by layer, using this project's own numbers

- **Landowners**: already realising gains today. EcoWorld's single Kulai land sale (Notebook 7) is expected to lift its FY2027 profit by 18-26% on its own, real, already-recognised value.
- **TNB (the grid)**: funding a RM42.8 billion three-year capex programme (Notebook 4) partly through tariff increases on the very customers driving that demand; TNB captures value through pricing, not equity ownership, and its exposure grows with connected capacity regardless of whether that capacity is ever fully utilised.
- **Contractors**: real, contracted revenue today, but highly uneven by company. Sunway Construction's data centre exposure is large relative to its whole business; Gamuda's is a small slice of a much bigger, more diversified book (Notebook 6). None of the three contractors' own guidance supports a distinctly higher margin for this work versus their normal business.
- **YTL Power**: the one operator with fully realised, audited profit from actually running data centres, RM345.5 million profit before tax in FY2026 (Notebook 5), though still a small share of its total group revenue, and earned in a year when the rest of the group's profitability fell sharply.
- **Hyperscalers**: the layer with the least verifiable value capture today, over $23 billion in combined pledges (Notebook 8), but no public disbursement data for any of them, meaning their real financial exposure to Malaysia specifically cannot currently be measured from public information.

Ranked by how *verifiably realised* (not necessarily how large) each layer's value capture is today: landowners and YTL Power have the most concrete, already-recognised gains; contractors have real revenue but thinner, contested margins; TNB's capex is real spending funded by real tariff increases, with returns realised over years rather than immediately; hyperscalers' position is the least verifiable of all, real pledges, unknown disbursement.

## A genuine wage premium, but not many jobs

Bank Negara Malaysia's own data shows data centre jobs pay well above the national median: entry-level roles at **RM3,000-7,000 per month**, senior roles at **RM10,000-30,000 per month**, against a national median salary across all sectors of **RM2,793 per month (2024)**. **[PUBLIC, Bank Negara Malaysia Q3 2025 Bulletin, Chart 5, sourced to DOSM and Jobstreet]** BNM's own commentary explicitly notes, however, that data centre job **volume** remains low compared to manufacturing: these are highly automated, capital-intensive facilities that employ relatively few permanent staff per megawatt of capacity built. The wage premium for the jobs that do exist is real; a large-scale employment story is not supported by the same data.

## Explain It Simply

If you follow the money through this whole ecosystem, from the moment land gets sold to the moment a hyperscaler actually pays for server capacity, some people are clearly getting paid today, and some are still just holding a promise. Landowners who sold at the right time and place have already banked real gains. The one company that actually runs a data centre business with audited numbers, YTL Power, is making real, growing profit from it, though it is still a small part of its overall business. Construction companies are earning real revenue building these facilities, though not at the dramatically higher profit margins many assume. TNB, the electricity company, is spending real money upgrading the grid and recovering that cost through higher bills. The tech giants who announced the biggest headline numbers, tens of billions of dollars, are, from the outside, the hardest to actually verify: nobody outside those companies can currently check how much of their promised money has actually shown up. And despite all this investment, the actual number of new jobs created is modest, even though the jobs that do exist pay well above Malaysia's typical salary.

## Glossary

- **Value capture**: which participant in an economic activity actually ends up with the financial benefit, as distinct from who is merely associated with the activity's headlines.
- **GDP contribution**: the portion of a country's total economic output (Gross Domestic Product) attributable to a specific activity or sector.

## Why Is This Happening?

**Why does value capture look so uneven across this ecosystem?** Each layer bears a different amount of risk and commits capital on a different timeline. Landowners take on essentially no ongoing risk once a sale completes, they are paid immediately and fully, which is exactly why their gains show up first and most clearly. Contractors take on execution risk (cost overruns, delays) for a fixed contract price, a moderate risk profile that explains their moderate, competitively-bid margins. YTL Power, as the actual long-term operator, takes on the largest ongoing risk, occupancy risk, technology risk, multi-decade capital commitment, and is compensated with a genuinely growing profit stream once facilities are operating, exactly the segment-level pattern Notebook 5 found. Hyperscalers' pledges sit furthest from realised value capture in this specific market because their financial exposure is diversified globally and reported at a corporate level far removed from any single country, which is also precisely why Malaysia-specific disbursement data doesn't publicly exist for them.
