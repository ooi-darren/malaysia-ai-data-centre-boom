## The scale of the demand

A 100MW data centre facility using evaporative cooling consumes roughly **4.16 million litres of water per day**, equivalent to the daily domestic use of around 10,000-18,000 Malaysian households, a figure consistent across multiple industry sources, though not traceable to a single primary engineering study. **[Well-corroborated secondary estimate, not a company- or government-stated figure]** Scaled up, cooling roughly 48 operational data centres by 2030 could require an estimated **675 million cubic metres of water per year**. **[w.media special report, an industry trade publication, not a primary government figure]**

Bank Negara Malaysia's own, more conservative calculation puts a 50MW facility's water use at roughly the equivalent of **2,200 households per day**, alongside an electricity-use equivalent of 22,000 households per day, sourced to DOSM per-capita consumption data and MIDA's own Water Usage Effectiveness guideline. **[PUBLIC, Bank Negara Malaysia Q3 2025 Bulletin]**

## A real, technical intervention, not a blanket moratorium

Johor has already acted, but more narrowly than a simple "moratorium" framing suggests. In **November 2025, Johor banned Tier 1 and 2 data centres outright**, the older, less water-efficient facility classifications, specifically because they use excessive water relative to higher-tier, more efficient designs. Separately, **facilities using water-intensive evaporative cooling systems have been told to wait until 2027** for new water-supply approval, tied to two new treatment plants coming online (the Semanggar Plant, 50 million litres/day capacity, expected 2026, and Layang 2 Phase 2, 160 million litres/day, expected 2027). Johor has also **raised industrial water tariffs for high-volume users since August 2025**. **[PUBLIC, w.media reporting, November 2025]** This is a real supply-capacity constraint with a specific, dated technical cause and fix, not an indefinite freeze on all data centres.

Part of the response is already operating: **12 million litres per day of treated effluent** from 13 waste treatment plants is being supplied to data centres by Indah Water Konsortium and Johor Special Water, a genuine mitigation already in place rather than a future promise. **[PUBLIC, w.media reporting, sourced to Indah Water Konsortium and Johor Special Water]**

## A real community impact, not just a technical one

On **8 February 2026, more than 50 residents representing roughly 1,000 people across four housing estates in Gelang Patah protested a data centre under construction for China's Zdata Technologies**, less than one kilometre from their homes. Their stated demands were an end to dust pollution and compensation for health impacts, with water supply also raised as a concern. The protest dispersed after about 90 minutes under police watch, and the developer did not meet with protesters. **[PUBLIC, South China Morning Post, a primary-grade independent news source reporting directly on the event]** This is the clearest documented instance in this project's research of the boom's physical footprint creating a direct, adversarial relationship with a specific local community, not an abstract resource-planning question.

## Explain It Simply

Cooling a data centre takes a huge amount of water, one medium-sized facility can use as much water in a day as thousands of households. Johor has already had to step in: it banned the oldest, least water-efficient type of data centre outright, and told any facility that wants to use the thirstiest kind of cooling system to wait until 2027, when new water treatment plants are ready. In the meantime, some of that water pressure is already being eased by recycling treated wastewater instead of using fresh water. This isn't just a spreadsheet problem, either, in February 2026 a group of residents in Gelang Patah protested a data centre being built right next to their homes, worried about dust and their health, a real example of this boom's physical costs landing on real people, not just abstract water-planning statistics.

## Glossary

- **Evaporative cooling**: a data centre cooling method that uses water evaporation to remove heat, effective and relatively cheap, but water-intensive compared to closed-loop or liquid cooling systems.
- **Tier classification**: an industry standard rating a data centre's design redundancy and reliability, from Tier 1 (basic) to Tier 4 (highest); lower tiers are generally older and less resource-efficient.
- **Treated effluent**: wastewater that has been processed to a quality suitable for non-drinking industrial reuse, such as data centre cooling.

## Why Is This Happening?

**Why would Johor ban older facility tiers rather than simply limiting total data centre numbers?** A blanket cap on the number of facilities would penalise newer, more water-efficient designs exactly as much as older, wasteful ones, discouraging the kind of technology upgrade that actually solves the underlying resource problem. Targeting the ban at Tier 1 and 2 facilities specifically incentivises new investment to arrive in the more water-efficient higher tiers, while the temporary wait for evaporative-cooling approvals buys time for water infrastructure (the two new treatment plants) to catch up, a more surgical policy response than an indiscriminate moratorium would have been, consistent with the same government instinct already found in Notebook 3's informal application filtering, manage growth rather than halt it outright.
